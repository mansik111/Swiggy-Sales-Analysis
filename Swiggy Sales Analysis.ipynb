### Daily Sales Trend

```python
df["DayName"] = df["Order Date"].dt.day_name()

daily_revenue = (
    df.groupby("DayName")["Price (INR)"]
    .sum()
    .reindex(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
)

plt.figure(figsize=(10, 5))
bars = plt.bar(daily_revenue.index, daily_revenue.values, color="#B19CD9")

# Add values on top of each bar
for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"₹{value:,.0f}",
        ha="center",
        va="bottom"
    )
    
plt.title("Revenue by Day of Week")
plt.xlabel("Day")
plt.ylabel("Revenue (INR)")

plt.tight_layout()
plt.show()
```

### Total Sales by Food Type (Veg vs Non-Veg)

```python
non_veg_keywords = ["chicken", "egg", "fish", 
                    "mutton", "prawn", "biryani", 
                    "kabab", "kebab", "non-veg", "non veg"]

df["Food Category"] = np.where(
    df["Dish Name"].str.lower().str.contains("|".join(non_veg_keywords), na=False),
    "Non-Veg",
    "Veg"
    
)
```

```python
food_revenue = (
    df.groupby("Food Category")["Price (INR)"]
    .sum()
    .reset_index()
)
```

```python
fig = px.pie(
    food_revenue,
    values="Price (INR)",
    names="Food Category",
    hole=0.5,
    title="Revenue Contribution: Veg vs Non-Veg",
    color="Food Category",
    color_discrete_sequence=["#FF7F7F","#90EE90"]
)
fig.update_traces(
    textinfo="percent+label",
    pull=[0.05,0]
)
fig.update_layout(
    height=500,
    margin=dict(t=60, b=40, l=40, r=40)
)
fig.show()
```

### Total Sales by State (Map Visualization)

```python
fig = px.bar(
    df.groupby("State", as_index=False)["Price (INR)"]
      .sum()
      .sort_values("Price (INR)", ascending=False),
    x="Price (INR)",
    y="State",
    orientation="h",
    title="Revenue by State (INR)",
)
fig.update_traces(marker_color="purple")

fig.update_layout(
    height=600,
    yaxis=dict(autorange="reversed")
)

fig.show()
```

### Quarterly Performance Summary

```python
df["Order_Date"] = pd.to_datetime(df["Order Date"])
df["Quarter"] = df["Order_Date"].dt.to_period("Q").astype(str)
quarterly_summary = (
    df.groupby("Quarter", as_index=False)
    .agg(
        Total_Sales=("Price (INR)", "sum"),
        Avg_Ratings=("Rating", "mean"),
        Total_Orders=("Order_Date", "count")
    )
    .sort_values("Quarter")
)

quarterly_summary
```

### Top 5 Cities by Sales

```python
top_5_cities = (
    df.groupby("City")["Price (INR)"]
    .sum()
    .nlargest(5)
    .sort_values()
    .reset_index()
)

fig = px.bar(
    top_5_cities,
    x="Price (INR)",
    y="City",
    orientation="h",
    title="Top 5 Cities by Sales (INR)",
    color_discrete_sequence=[""]
)

fig.show()
```

### Weekly Trend Analysis

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Week"] = df["Order Date"].dt.to_period("W").astype(str)

weekly_summary = (
    df.groupby("Week", as_index=False)
      .agg(
          Total_Sales=("Price (INR)", "sum"),
          Avg_Rating=("Rating", "mean"),
          Total_Orders=("Order Date", "count")
      )
      .sort_values("Week")
)
fig = px.line(
    weekly_summary,
    x="Week",
    y="Total_Sales",
    markers=True,
    title="Weekly Revenue Trend"
)

fig.update_layout(
    height=500,
    xaxis_title="Week",
    yaxis_title="Revenue (INR)"
)

fig.show()
```
